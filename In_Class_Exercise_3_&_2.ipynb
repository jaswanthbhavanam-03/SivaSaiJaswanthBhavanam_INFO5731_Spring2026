**Q2)** What are the top 3 most frequent companies? **(2 points)**

```python
# Starter cell (optional)
import requests
import pandas as pd

API_URL = "https://remoteok.com/api"
```

```python
# Write your solution below.
# Tip: the API response includes a metadata object plus job objects.
# Make sure you keep only real job postings.
import requests
import pandas as pd

API_URL = "https://remoteok.com/api"

# 1) Fetch JSON from API (NO selenium)
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json",
}

resp = requests.get(API_URL, headers=headers, timeout=30)
resp.raise_for_status()
data = resp.json()

# The API returns a list where the first element is metadata (like "legal"/"last_updated")
# Job postings are the objects that contain typical job fields like "position" and "company".
jobs = [
    item for item in data
    if isinstance(item, dict) and ("position" in item) and ("company" in item)
]

# -------------------------
# Q1) How many job postings?
# -------------------------
q1_total_jobs = len(jobs)

# ----------------------------------------
# Q2) Top 3 most frequent companies
# ----------------------------------------
companies = pd.Series([job.get("company") for job in jobs])
top3_companies = companies.value_counts().head(3)

# ------------------------------------------------------
# Q3) How many job titles contain the word "Data"?
#     (case-insensitive)
# ------------------------------------------------------
q3_data_titles = sum(
    "data" in str(job.get("position", "")).lower()
    for job in jobs
)

# ----------------------------------------
# Q4) DataFrame: first 10 job postings
# Columns:
#   title, company, location ("Unknown" if missing),
#   salary (None if missing), tags (comma-separated)
# ----------------------------------------
rows = []
for job in jobs[:10]:
    title = job.get("position")
    company = job.get("company")

    location = job.get("location")
    if not location or str(location).strip() == "":
        location = "Unknown"

    # Salary handling:
    # RemoteOK often provides salary_min/salary_max. If both are > 0, store "min-max".
    # Otherwise None.
    salary_min = job.get("salary_min", 0) or 0
    salary_max = job.get("salary_max", 0) or 0
    salary = f"{salary_min}-{salary_max}" if (salary_min > 0 and salary_max > 0) else None

    tags_list = job.get("tags", [])
    if isinstance(tags_list, list):
        tags = ", ".join(map(str, tags_list))
    else:
        tags = ""  # fallback

    rows.append({
        "title": title,
        "company": company,
        "location": location,
        "salary": salary,
        "tags": tags
    })

df = pd.DataFrame(rows)

# -------------------------
# Print results clearly
# -------------------------
print("===== Q1 =====")
print("Total job postings currently available:", q1_total_jobs)

print("\n===== Q2 =====")
print("Top 3 most frequent companies:")
print(top3_companies)

print("\n===== Q3 =====")
print('Number of job titles containing "Data" (case-insensitive):', q3_data_titles)

print("\n===== Q4 (df.head(10)) =====")
print(df.head(10))













```

 Q2( 4 points)
Define a base class called 'Vehicle' with attributes make and model. Create a derived class Car that inherits from Vehicle and has an additional attribute 'num_doors'. Demonstrate an example of creating an instance of the 'Car' class and accessing its attributes.

```python
# write your answer here
# Q2 Answer

class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model


class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors = num_doors


# Creating an instance of Car
my_car = Car("Hyundai", "Sonata", 5)

# Accessing attributes
print("Make:", my_car.make)
print("Model:", my_car.model)
print("Number of doors:", my_car.num_doors)

```

Question 3 ( 2 points)
Create a program that accepts a list of numbers as input and outputs a new list containing only the even numbers.

```python
# write your answer here
# Q3 Answer

numbers = input("Enter numbers separated by space: ")

num_list = list(map(int, numbers.split()))

even_numbers = [num for num in num_list if num % 2 == 0]

print("Even numbers:", even_numbers)

```
