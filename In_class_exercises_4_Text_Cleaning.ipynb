Load the TAB-separated file into a pandas DataFrame with columns: `id`, `message`.

Using RegEx, extract and create two new columns:

```python
# Setup (run this cell first)
import re
import pandas as pd
```

## Q1 (1 point) — Answer below

```python
# Q1 — ANSWER CELL

# file path
FILE_PATH = "/content/support_messages.txt"

# load TAB-separated file
df = pd.read_csv(FILE_PATH, sep="\t")

# print shape
print(df.shape)

# print first 3 rows
print(df.head(3))
```

## Q2 (3 points) — Answer below

```python
# Q2 — ANSWER CELL

# word_count → count words
df["word_count"] = df["message"].astype(str).str.split().str.len()

# char_count → count characters
df["char_count"] = df["message"].astype(str).str.len()

# num_count → count digits (0–9)
df["num_count"] = df["message"].astype(str).str.count(r"\d")

# upper_word_count → count ALL CAPS words
def count_upper_words(text):
    tokens = str(text).split()
    count = 0
    for token in tokens:
        if token.isupper():
            count += 1
    return count

df["upper_word_count"] = df["message"].apply(count_upper_words)

# print full dataframe
print(df)
```

## Q3 (3 points) — Answer below

```python
# Q3 — ANSWER CELL

# load stopwords (easy method)
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
STOPWORDS = set(ENGLISH_STOP_WORDS)

def clean_text(text: str) -> str:

    # make sure text is string
    text = str(text)

    # STEP 1: lowercase
    text = text.lower()

    # STEP 2: remove punctuation/symbols
    # keep only letters, numbers, spaces
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # STEP 3: remove stopwords
    words = text.split()
    words = [w for w in words if w not in STOPWORDS]
    text = " ".join(words)

    # STEP 4: remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


# create clean column
df["clean"] = df["message"].apply(clean_text)


# print original and cleaned text for id = 1 and id = 4
for target_id in [1, 4]:
    row = df[df["id"] == target_id].iloc[0]
    print("ID:", row["id"])
    print("ORIGINAL:", row["message"])
    print("CLEAN:", row["clean"])
    print("-" * 40)
```

## Q4 (2 points) — Answer below

```python
# Q4 — ANSWER CELL

# create regex patterns
order_pattern = re.compile(r"ORD-\d{4}", re.IGNORECASE)
email_pattern = re.compile(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}")

# function to extract order id
def extract_order_id(text):
    match = order_pattern.search(str(text))
    if match:
        return match.group(0).upper()   # make consistent format
    return None

# function to extract email
def extract_email(text):
    match = email_pattern.search(str(text))
    if match:
        return match.group(0)
    return None

# create new columns
df["order_id"] = df["message"].apply(extract_order_id)
df["email"] = df["message"].apply(extract_email)

# print required columns
print(df[["id", "order_id", "email"]])
```

## Q5 (1 point) — Answer below

```python
# Q5 — ANSWER CELL

from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# 1) create TF-IDF model
vectorizer = TfidfVectorizer()

# fit and transform using clean column
X = vectorizer.fit_transform(df["clean"])

# 2) compute average TF-IDF score per word
avg_tfidf = np.asarray(X.mean(axis=0)).ravel()

# get words (feature names)
terms = np.array(vectorizer.get_feature_names_out())

# 3) get indices of top 5 highest scores
top5_idx = avg_tfidf.argsort()[::-1][:5]

# 4) print top 5 keywords with scores
for i in top5_idx:
    print(terms[i], round(avg_tfidf[i], 4))
```
